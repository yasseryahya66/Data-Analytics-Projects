```python
# check the percentage of missing values
print(cars_data.isna().mean()*100)

# Many columns have major missing values
# color_slug should be dropped: 94 % missing values
# stk_year should be dropped: 84 % missing values
```

```python
# drop color_slug and stk_year columns from the dataset
cars_data = cars_data.drop(columns=['color_slug','stk_year'])
```

```python
print(cars_data.info())
```

```python
# check the missing values for 'maker' column
print(cars_data['maker'].isna().sum())
print(cars_data['maker'].isna().mean()*100)

# 14 % are missing of maker variable
# for each missing value in maker, there is a missing one in model
# I would drop nan values in 'maker'
```

```python
# drop nan values from 'maker'
cars_data = cars_data.dropna(subset=['maker'])

# check missing values
print(cars_data['maker'].isna().sum())
print(cars_data['maker'].isna().mean()*100)
```

```python
# check missing values for 'model' column
print(cars_data['model'].isna().sum())
print(cars_data['model'].isna().mean()*100)

# the number of missing values for model are reduced to 20 %
# fill missing model using most common model per maker
```

```python
# fill missing model using most common model per maker
cars_data['model'] = cars_data['model'].fillna(
    cars_data.groupby('maker')['model']
    .transform(lambda x: x.mode()[0] if not x.mode().empty else None)
)
```

```python
# check missing values for 'model' column
print(cars_data['model'].isna().sum())
print(cars_data['model'].isna().mean()*100)
```

```python
# convert data type from objects into strings for better pandas handling
cars_data['maker'] = cars_data['maker'].astype('string')
cars_data['model'] = cars_data['model'].astype('string')

print(cars_data['maker'].dtype)
print(cars_data['model'].dtype)
```

```python
print(cars_data['mileage'].dtype)
print(cars_data['mileage'].isna().sum())
print(cars_data['mileage'].isna().mean()*100)

# mileage is a numeric data type'float'
# we could fill missing values using the 'median'
```

```python
cars_data['mileage'].median()
```

```python
# fill missing values for 'mileage' using median
cars_data['mileage'] = cars_data['mileage'].fillna(cars_data['mileage'].median())

# check missing data again
print(cars_data['mileage'].isna().sum())
print(cars_data['mileage'].isna().mean()*100)
```

```python
# check missing values of manufacture_year
cars_data['manufacture_year'].isna().mean()*100

# the percentage of missing values is around 10 %
# we fill using median year of each maker group
```

```python
# Fill missing manufacture_year values using the median year of each maker group
# (cars from the same maker tend to have similar production years, so we use group-based median)
cars_data['manufacture_year'] = cars_data['manufacture_year'].fillna(
    cars_data.groupby('maker')['manufacture_year'].transform('median')
)

# check again if missing values are filled by median
print(cars_data['manufacture_year'].isna().mean()*100
)
```

```python
# convert the data type from float into integer
cars_data['manufacture_year'] = cars_data['manufacture_year'].astype(int)

# check the data type again
cars_data['manufacture_year'].dtype
```

```python
# Check outliers
# List of numerical columns
columns = [
    'price_eur',
    'mileage',
    'engine_power',
    'engine_displacement'
]

# Loop through each column
for column in columns:

    # Calculate Q1 (25th percentile)
    Q1 = cars_data[column].quantile(0.25)

    # Calculate Q3 (75th percentile)
    Q3 = cars_data[column].quantile(0.75)

    # Calculate IQR
    IQR = Q3 - Q1

    # Define lower boundary
    lower_bound = Q1 - 1.5 * IQR

    # Define upper boundary
    upper_bound = Q3 + 1.5 * IQR

    # Find outliers
    outliers = cars_data[
        (cars_data[column] < lower_bound) |
        (cars_data[column] > upper_bound)
    ]

    # Print results
    print(f'\n{column}')
    print(f'Number of outliers: {len(outliers)}')
```

```python
# delete outliers
cars_data = cars_data[
    (cars_data[column] >= lower_bound) &
    (cars_data[column] <= upper_bound)
]
```

```python
# List of numerical columns to visualize on boxplot
columns = [
    'price_eur',
    'mileage',
    'engine_power',
    'engine_displacement'
]

# Loop through each column in the list
for column in columns:

    plt.figure(figsize=(15,10))

    sns.boxplot(
        y='maker',
        x=column,
        data=cars_data
    )

    plt.title(f'{column} by Maker')

    plt.show()
```

```python
# Plot bar plots for some categorical column
for column in ['maker', 'body_type', 'transmission', 'door_count', 'seat_count', 'fuel_type']:
    plt.figure(figsize=(8,5))
    cars_data[column].value_counts().plot(kind='bar')
    plt.title(column)
    plt.show()
```
